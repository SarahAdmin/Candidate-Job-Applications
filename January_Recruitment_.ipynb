```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
jandata = pd.read_csv('Jan Feedback.csv')
```

```python
janStage = jandata['Current Stage'].value_counts()
```

```python
janStage
```

```python
janStage.plot(kind='pie',autopct='%1.1f%%',title='January Application Current Stage',ylabel='')
```

```python
janArchive = jandata['Application Archive Reason'].value_counts()
```

```python
janArchive
```

```python
janArchive.plot(kind='pie',autopct='%1.1f%%',title='January Application Archive Reasons',ylabel='')
```

```python
stage1 = sns.displot(data=jandata,x='Current Stage',hue='Posting Title',multiple='stack')
stage1.set(ylabel='Number of Applications')
plt.xticks(rotation=80)
```

```python
archive1 = sns.displot(data=jandata,x='Application Archive Reason',hue='Posting Title',multiple='stack')
archive1.set(ylabel='Number of Applications')
plt.xticks(rotation=80)
```
