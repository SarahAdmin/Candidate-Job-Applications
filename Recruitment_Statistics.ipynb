```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
septdata = pd.read_csv('September Feedback.csv')
```

```python
septdata.info()
```

```python
septdata['Current Stage']
```

```python
septdata['Application Archive Reason']
```

```python
appArchive = septdata['Application Archive Reason'].value_counts()
```

```python
appArchive
```

```python
appArchive.plot(kind='pie',title='September Application Archieved',autopct='%1.1f%%',ylabel='Number of Applications')
```

```python
appStage = septdata['Current Stage'].value_counts()
```

```python
appStage
```

```python
appStage.plot(kind='pie',autopct='%1.1f%%',title='September Application Current Stage',ylabel='Number of Applications')
```

```python
stage1 = sns.displot(data=septdata,x='Current Stage',hue='Posting Title',multiple='stack')
stage1.set(ylabel='Number of Applications')
plt.xticks(rotation=80)
```

```python
archive1 = sns.displot(data=septdata,x='Application Archive Reason',hue='Posting Title',multiple='stack')
archive1.set(ylabel='Number of Applications')
plt.xticks(rotation=80)
```

On the pie chart and bar graph, the most popular Application Archive Reason was 'Underqualified' because candidates has been rejected out applications as not moving forward at this time for various job roles.
On Current Stage charts, Hiring Manager Screen and Offer has same frequency.
